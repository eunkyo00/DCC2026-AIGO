```python
import json, sys
from collections import Counter
from pathlib import Path

MISSION_DIR = Path.cwd() if Path.cwd().name == 'mission-1' else Path.cwd() / 'mission-1'
PROJECT_ROOT = MISSION_DIR.parent
sys.path.insert(0, str(MISSION_DIR / 'baseline'))
sys.path.insert(0, str(PROJECT_ROOT))
from feature_extraction import build_or_load_features
from train_svm import run
```

## 1–4. Manifest 로드, 표본/성별 분포, MFCC feature shape

```python
train = build_or_load_features(PROJECT_ROOT/'manifests/mission1_gender_Training.csv', MISSION_DIR/'cache', 'train')
valid = build_or_load_features(PROJECT_ROOT/'manifests/mission1_gender_Validation.csv', MISSION_DIR/'cache', 'valid')
print('Training samples:', len(train.y), Counter(train.y))
print('Validation samples:', len(valid.y), Counter(valid.y))
print('Feature shapes:', train.X.shape, valid.X.shape)
```
